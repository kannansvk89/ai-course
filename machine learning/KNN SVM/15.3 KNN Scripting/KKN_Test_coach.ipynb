```python
import pandas as pd
from sklearn.model_selection import train_test_split,KFold,cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
from sklearn.datasets import load_wine

import warnings
warnings.filterwarnings("ignore")
```

## 1. Data Collection

```python
load_wine()
```

```python
wine_data = load_wine()
wine_data_df = pd.DataFrame(data = wine_data.data, columns=wine_data.feature_names)
wine_data_df['default'] = wine_data.target
wine_data_df
```

## 3. Model Building || 4. Model Training || 5. Model Testing || 6. Model Evalution

```python
X = wine_data_df.drop("default",axis=1)
y = wine_data_df["default"]

Xtrain,Xtest, ytrain, ytest = train_test_split(X,y,test_size=0.20,shuffle=True, stratify=y,random_state=123)
kmean_alg_classfier = KNeighborsClassifier(n_neighbors=3)
kmean_alg_classfier.fit(Xtrain,ytrain)
y_pred_train = kmean_alg_classfier.predict(Xtrain)
print("Accuracy Score:",accuracy_score(ytrain,y_pred_train))
```

## Since the each input features are in different units lets make it standardize it

```python
standard_scaler = StandardScaler()
scaled_X = standard_scaler.fit_transform(X)

Xtrain,Xtest, ytrain, ytest = train_test_split(scaled_X,y,test_size=0.20,shuffle=True, stratify=y,random_state=123)
kmean_alg_classfier.fit(Xtrain,ytrain)
y_pred_train = kmean_alg_classfier.predict(Xtrain)
print("Accuracy Score:",accuracy_score(ytrain,y_pred_train))
```

### Choose optimal k costant 

```python
kmean_scores = []
kfold = KFold(shuffle=True,n_splits = 5,random_state=123)
for i in range(3,52,2):
    kmodel = KNeighborsClassifier(n_neighbors=i)
    kmodel.fit(scaled_X,y)
    cvscores = cross_val_score(estimator=kmodel,X=scaled_X, y=y,cv = kfold)
    kmean_scores.append(cvscores.mean().round(2))

    cross_val_score()

kmean_scores.index(max(kmean_scores))
kmean_scores[5]
```
