# Handling Imbalanced Dataset
- Basically the given data set after understaning , the output Y is discrete
- Imbalance data only occur in classification
- During train test spilit using pie chart understand how y is equally distributed for both y_train and y_test , it should be same, this can achieved by stratify hyber param
- Also in a given dataset if our expected data distribution is less
-   example : in the breast cancer dataset 63% benign is there but only 37% Malignant is there
-   So, while model buildeing using class weights define the weightage, as this dataset 63:37, we can give Malignant weight - 2, Benign weight - 1

#### Import Necessary Libraies

```python
from sklearn import datasets
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import confusion_matrix, classification_report
from matplotlib import pyplot as plt
```

# 1. Business understanding

```python
print(datasets.load_breast_cancer())
```

# 2. Data Collection

```python
breast_cancer_dataset = datasets.load_breast_cancer()
breast_cancer_data_df = pd.DataFrame(breast_cancer_dataset.data,columns=breast_cancer_dataset.feature_names)
breast_cancer_data_df
```

```python
print(breast_cancer_dataset.DESCR)
```

```python
breast_cancer_data_df["Target"] = breast_cancer_dataset.target
breast_cancer_data_df
```

### Note:
- from the given data set understand that it is already distributed  212 - Malignant, 357 - Benign
- Malignant is danger if the person affected, here around 65% postive data 212 expected result as low,
- Becuase we need maligant, if we find malignant earlier we can try to save the person
- in a given dataset the data is imbalance 65% Benign and 35% malignant, so if we not effectively handle the data then result will fall on beign
- For that only we are going to give weightage during model training that

# 5. Model Building

### 1. Input and output separation

```python
x= breast_cancer_data_df.drop(labels="Target",axis = 1)
y= breast_cancer_data_df["Target"]
x.shape,y.shape
```

#### Apply Train test spilt

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,random_state = 1234,shuffle=True,stratify = y)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

### Analyze how the data is classified using pie chart

```python
plt.pie(x = y.value_counts(),labels=["Benign Cancer", "Malignant Cancer"],autopct="%1.0f%%")
plt.show()
```

```python
plt.pie(x = y_train.value_counts(),labels=["Benign Cancer", "Malignant Cancer"],autopct="%1.0f%%")
plt.show()
```

```python
plt.pie(x = y_test.value_counts(),labels=["Benign Cancer", "Malignant Cancer"],autopct="%1.0f%%")
plt.show()
```

#### Note:
- as train vs test data proportion not same in pie chart, we need to use hyper param straitify
- Actual x and Y showing 63 : 37
- But train showing 63: 37 and Test is 61 : 39
- To balance both train and test with same proportion , using stratify

# 6. Model Building

```python
dt_model = DecisionTreeClassifier(criterion='gini',max_depth=3,class_weight={0:2,1:1})
dt_model.fit(x_train,y_train)
```

#### Let's which algoirthm best suite for this dataset using grid search cv
- GridSearchCV itself does not decide which algorithm family is best like linear/logical/Decision Tree. It only tunes the parameters of the estimator you pass in.
- if you want check which algorithm can go for that you have to apply common sense or have the performance result all three and then choose the best

```python

grid_search = GridSearchCV(estimator  = dt_model,param_grid = {"max_depth":[5,6,7,8,9,10],
                                                                "criterion":['gini','entropy']})
grid_search.fit(x,y)

print(grid_search.best_params_)
print(grid_search.best_score_)
```

# 7. Model Training

```python
y_pred_train = dt_model.predict(x_train)
y_pred_train
```

# 8. Model Testing

```python
y_pred_test = dt_model.predict(x_test)
y_pred_test
```

# 9. Model Evaulation

#### Train data evaulation

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

#### Test Data Evaultion

```python
print(classification_report(y_test,y_pred_test))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```

#### Note: 
If you compare train vs test perfomance result , there is nearly 10% variance are there
 - This is due to imbalance dataset
- During model builidng we can analyze using chart how the M and B is distributed 
- So train and test data data distribuiton should be in same proportion as it is in the origninal data set
