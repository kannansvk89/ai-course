```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree
```

```python
df = pd.read_csv("home_buyers.csv")
```

```python
df
```

```python
X = df[["Income","Age","CIBIL","PreviousPurchase"]]
y = df["PurchaseHome"]
dt_model = DecisionTreeClassifier(   criterion="gini", 
    max_depth=3,          # allow deeper splits
    min_samples_split=2,  # allow splitting small groups
    random_state=42)
dt_model.fit(X,y)
```

```python
print(export_text(dt_model, feature_names=list(X.columns)))
```

```python
plt.figure( figsize = (8, 8),)
plot_tree(decision_tree = dt_model,fontsize=12,filled=True)
plt.show()
```
