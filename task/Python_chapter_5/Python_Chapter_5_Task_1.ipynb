```python
import seaborn as sns
import pandas as pd
from matplotlib import pyplot as plt
# List all available datasets
#print(sns.get_dataset_names())
# Load a dataset (e.g., 'tips')
tips_data = sns.load_dataset("tips")
#pd.set_option("display.max_rows", None)
tips_data.head()
```

#### 1. Historgram  - How tips are distributed

```python
plt.hist(tips_data.tip,color="skyblue", edgecolor="black")
plt.title("Histogram of tip")
plt.xlabel("Tips on the Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
#Insights
#1) Most of the customers are providing tips between 2 to 3 Rs
#2) The highest tips amount is 10 RS, but very few customers only provided
```

#### 2. Bar plot  - How avg tips amount is distributed between male and female customer

```python

sns.barplot(data = tips_data, x = "sex",y="tip")
plt.title("Average tip by sex")
plt.xlabel("Sex")
plt.ylabel("Tip")
plt.show()
```

```python
#Insights
#1) The average tip is provided by female is 2.8
#2) The average tip is provded by male is 3
#3) So compare to Female , male is providing more tips
```

#### 3. CountPlot - How size is distributed 

```python
sns.countplot(data = tips_data,x="size")
plt.title("Custome size distribution")
plt.show()
```

```python
#Insights
#1) Mostly customer with 2 heads are visting the restaraunt 
#2) Customers with more than 2 to 4 heads visting restaraunt is moderate
```

#### 4. Pie chart - how time distributed

```python
tips_data.value_counts("day").plot.pie(autopct="%1.1f%%")
plt.title("Day Proportion -  which day more count of biling happens")

plt.show()
```

```python
#Insights
#1) Looking at chart , on saturday and sunday more billing count is happening, since it is fall on week ends
```

#### 5. Scatter Plot

```python
sns.scatterplot(data=tips_data, x="total_bill", y="tip", hue="sex", size="size",)
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
#Insights
#1) I see if the bill amount is near 50 then tip amout falls between 8 to 10
#2) Compare to Female , Male customers are giving higher tip
# 3) Also, I see if customer heads are increated then tip amount also is collected reasonably
```

#### 6. Linechart - Analyze betweeen contionuse and discrete data

```python

# Sort by total_bill so the line makes sense
tips_sorted = tips_data.sort_values("total_bill")
plt.figure(figsize=(8,5))
sns.lineplot(x="total_bill", y="tip",data = tips_sorted, marker="o")
plt.title("Bill Vs Tip Analysis")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

```python
#insights
#1) If bill amounts are increasing then tip amount incereased
```

#### 7. Stacked chart

```python
pd.crosstab(tips_data["sex"], tips_data["smoker"]).plot(kind="bar", stacked=True)
plt.title("Stacked Bar: sex vs smokers")
plt.xlabel("Sex")
plt.ylabel("Smoker")
plt.show()
```

```python
#Insights
    #1) 60% of male not a smoker
    #2) In Female, out of 90 , 30% are smoker
```

#### 8. HeatMap

```python

tips_dropped = tips_data.drop(columns=['sex','day','time'])
# Compute correlation matrix (only numeric columns)
corr = tips_dropped.corr()
# Plot heatmap
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", linewidths=0.5)

plt.title("Correlation Heatmap - Tips Dataset")
plt.show()
```

```python
#Insights
#1) Strong Positive Correlation between total_bill ↔ tip As the bill amount increases, the tip amount also increases.
#2) Weak Correlation bertween size ↔ tip The number of people at the table doesn’t strongly determine the tip amount.
```
