## 1.1 List Built-in Functions

```python
#Append
list1 = [] #Declaring Empty List
list1.append("Kannan")
list1.append(66)
list1.append(True)
list1.append('N')
print(list1)
```

```python
#extend
budget_car = ["TATA","Maruti"]
cars = ['Ford', 'BMW', 'Volvo']
cars.extend(budget_car)
cars
```

```python
#index
budget_car = ["TATA","Maruti",'Ford', 'BMW', 'Volvo']
budget_car.index("Ford")
```

```python
#insert
budget_car = ["TATA","Maruti",'Ford', 'BMW', 'Volvo']
budget_car.insert(0,"Hyundai")
budget_car
budget_car.insert(-1,"Audi")
budget_car
```

```python
#Pop
budget_car = ["TATA","Maruti",'Ford', 'BMW', 'Volvo']
budget_car.pop()

budget_car.pop(-2)
budget_car
```

```python
#Remove
budget_car = ["TATA","Maruti",'Ford', 'BMW', 'Volvo']
budget_car.remove("Ford")
budget_car
```

```python
#Reverse
budget_car = ["TATA","Maruti",'Ford', 'BMW', 'Volvo']
budget_car.reverse()
budget_car
```

```python
#Sort
budget_car = ["audi","TATA","Maruti",'mord', 'BMW', 'Volvo']
budget_car.sort()
print(budget_car)
#Sort with case insensitive
budget_car.sort(key =str.lower)
print(budget_car)

#Sort with case insensitive and reverse
budget_car.sort(key =str.lower,reverse=True)
budget_car
```

## 1.2 Set Built-in Functions

```python
#add
set1 = set() #Declaring empty set
type(set1)
set1.add(True)
set1.add("kannan")
set1.add(45)
set1.add(45.56)
set1.add(45) ##Added duplicate
print(len(set1))
```

```python
#difference
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple","banana"}
z = x.difference(y)  ##Returns only the first set elements which not there in second(y) 
#This also can be done x - y
print(z)

#Ex
a = {"apple", "banana", "cherry"}
b = {"google", "microsoft", "apple"}
c = {"cherry", "micra", "bluebird"}

myset = a - b - c

print(myset)
```

```python
#difference upate - that modifies the original set

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple","banana"}
set1.difference_update(set2)  
print(set1)
```

```python
#Discard
set1 = {"apple", "banana", "cherry"}
set1.discard("apple")
set1
```

```python
#intersection - 
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple","banana"}
print(set1.intersection(set2))
#intersection_udpate will change originial set 1
```

```python
#disjoint - is both sets are different in across elements
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple","banana"}
print(set1.isdisjoint(set2))
```

```python
#issuperset - Returns True if all items of another set is present in this set
set1 = {"apple", "banana", "cherry","google","microsoft","IBM"}
set2 = {"google", "microsoft", "apple","banana","cherry"}
print(set1.issuperset(set2)) # can use symbol also ex : set1 >= set2

print(set1 > set2) ## that means set1 should have extra elements



```

```python
#pop - The pop() method removes a random item from the set.
set1 = {"apple", "banana", "cherry","google","microsoft"} 
set1.pop()
print(set1)
```

```python
#remove - specified element
set1 = {"apple", "banana", "cherry","google","microsoft"} 
set1.remove("apple")
set1
```

```python
#Symmetric - Return a set that contains all items from both sets, except items that are present in both sets:
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple","banana"}
print(set1.symmetric_difference(set2)) # can use symbol also ex : set1 ^ set2
```

```python
#unions
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple","banana"}
print(set1.union(set2)) # can use symbol also ex : set1 | set2
```

```python
#Subset
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple","banana","cherry"}
set1.issubset(set2) #can also use set1 < set2
print(set1<=set2)

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple","banana","cherry"}
print(set2 < set1)


```

```python
#update - it is same like extend a set by another set
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple","banana","cherry"}
set1.update(set2)   # can use symbol also ex : set1 |= set2
set1
set1.update(["ABC"]) #Set can be exted from list
print(set1)
set1.update({"dict"}) # set can be extend from antoher set
print(set1)
set3 = set(list1)
print(set3)
```

```python
#Creating tuple from dict
d = {'apple': 1, 'banana': 2, 'cherry': 3}
tup = tuple(d.items()) # Convert tuple of tuples
print(tup)
type(tup)
print(tup[0:1])
```

```python
#length
tup = (1, 2, 3, 4, "kannan",True)
print(len(tup))
```

## 1.4 Dictonary Built-in Functions

```python
#Delete a key
#initializing dict 
dict = {"name":"Kannan","Hobby":"Learn new things","age":36,"fav-fruit":["mango","banana"]}
print(type(dict))
print(dict["fav-fruit"][0]) ##accessing the elements
dict["name"] = "N Kannan" ##modifying
print(dict)
del dict["age"] ##remove a key
print(dict)
```

```python
#pop removes the item with the given key and returns its value
d = {"a": 1, "b": 2}
val = d.pop("a")
print(val)
print(d)
```

```python
#popitem  removes and returns the last inserted key-value pair
d = {"a": 1, "b": 2}
print(d.popitem())
```

```python
#clear -remove all item
d = {"a": 1, "b": 2}
d.clear()
print(d)
```

```python
#iterating dict - by key
d = {"a": 1, "b": 2}
for key in d:
    print(key)
```

```python
#iterating dict  - by value
d = {"a": 1, "b": 2}
for value in d.values():
    print(value)
```

```python
#iterae key and value together
d = {"a": 1, "b": 2}
for key, value in d.items():
    print(key, value)
```

```python
#copy
c = d.copy()
print(c)
```

```python
#From keys - Making new dictionary with arrays of key and set default value
keys = ["Math", "Science", "English"]
marks = dict.fromkeys(keys, 0) 
print(marks)
```

```python
#setDefault
#If the key exists → returns its value.
#If the key doesn't exist → inserts the key with the default value and returns that value.
student = {"name": "John"}
print(student.setdefault("age", 18))

print(student)
```

```python
#update - add new key value pair
student = {
    "name": "John"
}
student.update({"age": 20}) ## if key already exist it will simply override
print(student)
```
