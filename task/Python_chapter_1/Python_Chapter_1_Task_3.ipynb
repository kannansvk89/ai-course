```python
import random
words = ["Rock","Paper","Scissor"]
rock = "rock"
paper = "paper"
scissors = "scissor"
a = {"rockscissor":"rock","rockpaper":"paper","scissorpaper":"scissor"}

print("------------------Welcome to Rock, Paper, Scissors Game-------------------")
player_1 = "System"
player_2 = input("Enter your name")
print("Hi",player_2,"you are going to play with your computer as opponent")
print("\nGame Starts Now ......")

player_1_action = random.choice(words).lower()
player_2_action = input("You are player_2 select one of these Rock\Paper\Scissor :").lower()


iterate = True
while iterate:
    if player_2_action.capitalize() not in words:
        player_2_action = input("player_2 please, select one of these Rock\Paper\Scissor :").lower()
        continue
    print("Player_1 selected action : ",player_1_action)
    if player_1_action == player_2_action:
        print("Match draw")
    elif a.get(player_1_action+player_2_action) == rock or a.get(player_2_action+player_1_action) == rock:
        if player_1_action == rock:
            print("Player_1",player_1, "is won the match")
        else:
            print("Player_2",player_2, "is won the match:")
    elif a.get(player_1_action+player_2_action) == paper or a.get(player_2_action+player_1_action) == paper:
        if player_1_action == paper:
            print("Player_1",player_1, "is won the match")
        else:
            print("Player_2",player_2, "is won the match")
    elif a.get(player_1_action+player_2_action) == scissors or a.get(player_2_action+player_1_action) == scissors:
        if player_1_action == scissors:
            print("Player_1",player_1, "is won the match")
        else:
            
            print("Player_2",player_2, "is won the match")
    break
print("!!!Game End")
```
