# 1. Word Guessing Game (Mini Hangman)

```python
import random
print("----------------------Hi, Greetings, Welcome to the Word Guessing Game---------------------")
words = ["tree","book","fish","lamp"]
selected_word = random.choice(words)

finding_word = ["_"] * len(selected_word)

print("\nYour guessing word is with 4 letter _ _ _ _   you are given max 6 attempt to findout the word")
print("\nAll the Very Best!!")

for i in range(0,6):
    try:
        user_input = input("\nEnter the guessing letter").lower()
        
        if len(user_input) != 1 or not user_input.isalpha():
            raise ValueError("\n***You must enter exactly one alphabet letter")
            
        if user_input in selected_word:    
            for j in range(0,len(selected_word)):
                if user_input == selected_word[j]:
                    finding_word[j] = user_input

            current = "".join(finding_word)
            if current == selected_word:
                print("!!!Congrats you hunt the word and the word is -->",current)
                break
            else:                   
                print("\n!!!Great ",current, current.count("_"), "more letter to hunt the word, But you have only {} attempt left".format(5-i))
        else:
            print("\nSorry, Please try again")
             
        if i==5:
            print("\n!!!Game End, You utilized all 6 attempt, the hunting word is --> "+selected_word)
    except ValueError as e:
        print(e)
    
```
