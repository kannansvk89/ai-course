# 1.Rock, Paper, Scissors Game

```python
import random
words = ["Rock","Paper","Scissors"]
rock = "rock"
paper = "paper"
scissors = "scissors"
a = {"rockscissors":"rock","rockpaper":"paper","scissorspaper":"scissors"}

print("------------------Welcome to Rock, Paper, Scissors Game-------------------")
player_1 = "System"
player_2 = input("Enter your name")
print("Hi",player_2,"you are going to play with your computer as opponent")
print("\nGame Starts Now ......")

player_1_action = random.choice(words).lower()
player_2_action = input("You are player_1 select one of these Rock\Paper\Scissors :").lower()
print("Player_1 selected action : ",player_1_action)

if player_1_action == player_2_action:
    print("Match draw")
elif a.get(player_1_action+player_2_action) == rock:
    if player_1_action == rock:
        print("Player_1",player_1, "is won the match")
    else:
        print("Player_2",player_2, "is won the match:")
elif a.get(player_1_action+player_2_action) == paper:
    if player_1_action == paper:
        print("Player_1",player_1, "is won the match")
    else:
        print("Player_2",player_2, "is won the match")
elif a.get(player_1_action+player_2_action) == scissors:
    if player_1_action == scissors:
        print("Player_1",player_1, "is won the match")
    else:
        print("Player_2",player_2, "is won the match")

        





```
