# 1. Give 2 examples of all String functions which should be meaningful.

```python
#Center - padding
movie_title = input("Enter Movie title")
movie_title = movie_title.center(100,'-')
#Ex1
print(movie_title) 
#Ex2
print("The End".center(100)) 
```

```python
#count
user_input = input("Enter any words or line").lower()
#Ex1
print(user_input.count("india"))
#Ex2
print(user_input.count("in",13,30)) #search from staring point to end point
```

```python
#encode - default is utf-8
#Ex1
txt = "My name is Ståle"
print(txt.encode(encoding="ascii",errors="backslashreplace"))
print(txt.encode(encoding="ascii",errors="ignore"))
print(txt.encode(encoding="ascii",errors="namereplace"))
print(txt.encode(encoding="ascii",errors="replace"))
print(txt.encode(encoding="ascii",errors="xmlcharrefreplace"))
print("-----------------------------------------------------")
print(txt.encode(errors="backslashreplace"))
print("-----------------------------------------------------")
#Ex2 - Default encode
text = "தமிழ்"
print(text.encode())
```

```python
#Capitalize
text = "Kingkong";
print(text.capitalize())
```

```python
#Casefold
text = "iNdIa"
print(text.casefold())
```

```python
#Endswith
text = "Gen AI Agentic AI";
#Ex1
print(text.endswith("AI",8))
#Ex2
print(text.endswith("AI",8))
```

```python
#Expandtabs
text = "Gen AI Agentic \t AI";
#Ex1
print(text.expandtabs(30));
#Ex2 - default
print(text.expandtabs()); 
```

```python
#find ,rfind
text = "Gen AI Agentic AI";
#Ex1 - default
print(text.find("AI"))
#Ex2
print(text.rfind("AI",6,))

```

```python
#format
text = "My name firstname is {} and last name is {}";
#Ex1 - default
print(text.format("kannan","N"))
#Ex2
text = "My name lastname is {1} and first name is {0}";
print(text.format("kannan","N"))
#Ex3
txt = "We have {:^8} chickens." #added 8 space both left and right
print(txt.format(49))
#Ex4
#Use " " (a space) to insert a space before positive numbers and a minus sign before negative numbers:
txt = "The temperature is between {: } and {: } degrees celsius."
print(txt.format(-3, 7))
```

```python
#format_map
myvar = {"name" : "Ravi", "age" : 36}
txt = "Happy birthday {name} you are now on level {age}!"
print(txt.format_map(myvar))
```

```python
#index,rindex
text = "Gen AI Agentic AI";
#Ex1 - default
print(text.index("AI"))
#Ex2
print(text.rindex("AI",6,0)) ## index will raise an error is the substing not found, but find method simply return -1 as silent
```

```python
#isalnum
text = "Program"
print(text.isalnum())
```

```python
#isalpha
text = "Program1"
print(text.isalpha())
```

```python
#isascii
text = "Gen AI d! கீ";
print(text.isascii())
```

```python
#isdecimal - (0-9)
a = "\u0030" #unicode for 0
b = "\u0047" #unicode for G
c = "9"

print(a.isdecimal())
print(b.isdecimal())
print(c.isdecimal())
```

```python
#isidentifier
txt = "Demo"
x = txt.isidentifier()
print(x)
txt2 = "Demo 1"
print(txt2.isidentifier())
```

```python
#isprintable
txt = "Hello! \tAre you #1?"
x = txt.isprintable()
print(x)
```

```python
#istitle
txt = "The Gen Ai"
print(txt.istitle())

txt = "The Gen AI"
print(txt.istitle())
```

```python
#join
myDict = {"name": "John", "country": "Norway"}
mySeparator = ","
#Ex1
x = mySeparator.join(myDict) #To extract keys
print("json keys : ",x)
#Ex2
txt = ("Mango","Banana","Grapes")
print("#".join(txt))
```

```python
#ljust & rjust
#Ex1
txt = "banana"
x = txt.ljust(20)
print(x, "is my favorite fruit.")
#Ex2
txt = "banana"
x = txt.rjust(20,"-")
print(x, "is my favorite fruit.")
```

```python
#maketran
#Ex1
txt = "Hi Sam!"
x = "mSa"
y = "eJo"
mytable = str.maketrans(x, y) #it give the mapping table for later translate
print(txt.translate(mytable)) # now it translate using mapping table
#Ex2
txt = "Hello Sam!"
mytable = str.maketrans("S", "P")
print(txt.translate(mytable))
```

```python
#partition
#Ex1
txt = "I could eat bananas all day"
x = txt.partition("apples")
print(x)
#Ex2
txt = "I could eat bananas all day"
x = txt.partition("bananas")
print(x)
```

```python
#Replace
#Ex1
txt = "one one was a race horse, two two was one too."
x = txt.replace("one", "three")
print(x)
#Ex2
txt = "one one was a race horse, two two was one too."
x = txt.replace("one", "three", 2) ## replace only two times
print(x)
```

```python
#rfind
txt = "Hello, welcome to my world."
x = txt.rfind("e")
print(x)
```

```python
#Split
#Ex1
txt = "apple#banana#cherry#orange"
x = txt.split("#")
print(x)
#Ex2
txt = "apple#banana#cherry#orange"
x = txt.split("#",1)
print(x)
```

```python
#Splitlines
#Ex1
txt = "Thank you for the music\nWelcome to the jungle"
x = txt.splitlines()
print(x)
#Ex2
txt = "Thank you for the music\nWelcome to the jungle"
x = txt.splitlines(True)
print(x)
```

```python
#Swapcase
txt = "Hello My Name Is PETER"
x = txt.swapcase()
print(x)
```

```python
#title
txt = "hello my name Is PETER"
x = txt.title()
print(x)
```

```python
#Zfil
a = "hello"
b = "welcome to the jungle"
c = "10.000"

print(a.zfill(10))
print(b.zfill(10)) #if length is less than characters length, then nothing doing
print(c.zfill(10))
```
