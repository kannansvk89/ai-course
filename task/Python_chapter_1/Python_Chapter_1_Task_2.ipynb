```python
#While loop & Break
#Ex1
secret_word="mango"
print("---------------------Welcome to the word hunt program----------------------")
print("Find the fruit name --->  M_ng_")
while(True):
     usr_input = input("Enter you guessing word")
     if usr_input.lower() == secret_word:
        print("Congrats you found the fruit name")
        break;
print("---------------------------------------------------------------------------")

#Ex2
count = 5
while count > 0:
    print(count)
    count -= 1
```

```python
#for loop
#Ex1
txt = "**********"
for i in range(0,10):
    print(txt.replace("*","",i))
print("---------------------------------------------------------------------------")
#Ex2
for ch in "CAT":
    print(ch)

```

```python
#if..elif else
#Ex1
user_input = input("Enter your prefer course DataScience OR GenAI")
if user_input == "DataScience":
    print("Welcome to the DataScience Course")
elif user_input == "GenAI":
     print("Welcome to the GenAI Course")
else:
    print("Please select the right course")
    
print("---------------------------------------------------------------------------")
#Ex2
marks = 75
if marks >= 90:
    print("A")
elif marks >= 60:
    print("B")
else:
    print("C")
```

```python
#	user defined functions without for loop, if loop
#Ex1
def add(a, b):
    return a + b
print(add(2, 3))
print("---------------------------------------------------------------------------")
#Ex2
def greet(name):
    return "Hello " + name

print(greet("Sam"))
```

```python
#User-defined functions with for loop
#Ex1
def print_numbers():
    for i in range(3):
        print(i)
print_numbers()
print("---------------------------------------------------------------------------")
#Ex2
def show_letters(word):
    for ch in word:
        print(ch)
show_letters("CAT")
```

```python
#User-defined functions with for loop and if condition
#Ex1
def even_numbers():
    for i in range(1, 6):
        if i % 2 == 0:
            print(i)
even_numbers()

#Ex2
def vowels(text):
    for ch in text:
        if ch in "aeiou":
            print(ch)

vowels("hello")
```

```python
#Continue
#Ex1
for i in range(5):
    if i == 2:
        continue
    print(i)
print("---------------------------------------------------------------------------")
#Ex2
for ch in "python":
    if ch == "t":
        continue
    print(ch)
```

```python
#for loop with in operator
#Ex1
for ch in "ABC":
    print(ch)
print("---------------------------------------------------------------------------")
#Ex2
for n in [1, 2, 3]:
    print(n)
```

```python
#for loop with if statement
#Ex1
for i in range(5):
    if i > 2:
        print(i)
print("---------------------------------------------------------------------------")
#Ex2
for ch in "python":
    if ch == "o":
        print(ch)
```

```python
#for loop with not in operator
#Ex1
for ch in "python":
    if ch not in "aeiou":
        print(ch)
print("---------------------------------------------------------------------------")
#Ex2
for n in [1, 2, 3, 4]:
    if n not in [2, 4]:
        print(n)
```

```python
#. if with in operator
#Ex1
name = "python"
if "t" in name:
    print("Found")
print("---------------------------------------------------------------------------")

#Ex2
nums = [1, 2, 3]
if 2 in nums:
    print("Present")
```

```python
#if with not in operator
#Ex1
name = "python"
if "z" not in name:
    print("Not found")
print("---------------------------------------------------------------------------")
#Ex2
nums = [1, 2, 3]
if 5 not in nums:
    print("Missing")
```
