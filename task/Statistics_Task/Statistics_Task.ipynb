### 1)	A company claims that the average time taken to resolve customer tickets is 30 minutes. A random sample of 40 tickets shows a mean resolution time of 33 minutes with a standard deviation of 8 minutes. At the 5% level of significance, test whether the average resolution time has increased.

```python
#Hypothesis Testing Taken: One-Sample T-Test Mean

#Question: Is the average resolution time of a random sample of 40 tickets significantly different from the current average of 75 minutes?

#Data collected for random 40 tickets

#Hypothessis setup:
#H0:  μ = 30 Current process/statement of no change - There is no significant difference in the average resolution time
#H1:  μ > 30 Statement of change/New Process - There is a significant difference in the average resolution time

#Selected 1-Tailed Test:
#To test if the random sample of 40 tickets resolution average time resulting greater than current/expected average time 30 mins

#H1 Result consideration: Based on the test results, there is only a 5% risk of occurrence. However, this scenario is extremely rare and highly unlikely to occur in practice.
```

```python
from scipy import stats
import numpy as np

#Taking/collecting observations of 40 tickets resolution time
observations_40_tickets = np.random.normal(loc = 33, scale = 5.0, size = 40).astype(int)
observations_40_tickets

t_test_stats, p_val = stats.ttest_1samp(a=observations_40_tickets,popmean = 30,alternative = 'greater')

#As p_val result always will be 2-tailed , so took divide by 2
p_val = p_val/2

if p_val < 0.05:
    print("We reject H0")
else:
    print("We do not reject H0")
```

```python
#Conclusion: At 5% level of significance, We can reject the Null Hyposthesis(H0) and claim that the average resolution time has increased
```

### 2) The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves performance.


```python
# Hypothesis Testing Taken: One-Sample T-Test Mean

# Question: Has the new training course improved the average certification exam score 
# compared to the historical average of 65?

# Data collected for random 50 students trained under the new course

# Hypothesis setup:
# H0: μ = 65  (Null Hypothesis - No improvement, average score remains 65)
# H1: μ > 65  (Alternative Hypothesis - Improvement, average score is greater than 65)

# Selected 1-Tailed Test:
# To test if the random sample of 50 students trained under the new course 
# has an average score greater than the historical average of 65.

#At the 5% level of significance, we accept only a 5% chance of mistakenly concluding that the new course improves 
#performance when in reality it does not.

## Program
from scipy import stats
import numpy as np

# Simulating/collecting observations of 50 students' scores (mean ~68, std ~10)
observations_50_students = np.random.normal(loc=68, scale=10.0, size=50).astype(int)
print("Sample scores:", observations_50_students)

# Performing one-sample t-test against historical mean = 65
t_test_stats, p_val = stats.ttest_1samp(a=observations_50_students, popmean=65, alternative='greater')

# p-value is already one-tailed because we used alternative='greater'
print("t-statistic:", t_test_stats)
print("p-value:", p_val)

p_val = p_val/2

# Decision rule at 5% significance level
if p_val < 0.05:
    print("We reject H0")
    print("Conclusion: At 5% level of significance, the new course(H1) has improved performance.")
else:
    print("We do not reject H0")
    print("Conclusion: At 5% level of significance, there is no evidence of improvement.")
```

#### 3) A company claims that 60% of its users are satisfied with its product. In a survey of 200 users, 134 users report satisfaction. At the 5% level of significance, test whether customer satisfaction is higher than claimed

```python
# Hypothesis Testing Taken: One-Sample Z-Test for Proportion

# Question: Is customer satisfaction higher than the claimed 60%?

# Data collected from survey of 200 users
# 134 users reported satisfaction

# Hypothesis setup:
# H0: p = 0.60 (Null Hypothesis - No improvement, satisfaction is 60%)
# H1: p > 0.60 (Alternative Hypothesis - Satisfaction is higher than 60%)

# Selected 1-Tailed Test:
# To test if the proportion of satisfied users is greater than 0.60

from statsmodels.stats.proportion import proportions_ztest

# Sample data
n = 200
successes = 134
p0 = 0.60

# Perform one-sample z-test for proportion
z_stat, p_val = proportions_ztest(count=successes, nobs=n, value=p0, alternative='larger')


print("z-statistic:", z_stat.round(2))
print("p-value:", p_val.round(3))

p_val = p_val/2

# Decision rule at 5% significance level
if p_val < 0.05:
    print("We reject H0")
    print("Conclusion: At 5% level of significance, customer satisfaction (H1) is higher than 60%, So H1 is considered")
else:
    print("We do not reject H0")
    print("Conclusion: At 5% level of significance, there is no evidence that satisfaction is higher than 60%.")
```

### 4) Historically, 40% of applicants pass an entrance test. After a new preparation program, 90 out of 180 applicants pass. At the 5% level of significance, test whether the pass rate has improved.

```python
# Hypothesis Testing Taken: One-Sample Z-Test for Proportion

# Question: Has the new preparation program improved the pass rate 
# compared to the historical pass rate of 40%?

# Data collected from 180 applicants
# 90 applicants passed the test

# Hypothesis setup:
# H0: p = 0.40 (Null Hypothesis - No improvement, pass rate is 40%)
# H1: p > 0.40 (Alternative Hypothesis - Pass rate has improved)

# Selected 1-Tailed Test:
# To test if the proportion of applicants passing is greater than 0.40

from statsmodels.stats.proportion import proportions_ztest

# Sample data
n = 180
successes = 90
p0 = 0.40

# Perform one-sample z-test for proportion
z_stat, p_val = proportions_ztest(count=successes, nobs=n, value=p0, alternative='larger')

p_val = p_val/2
print("z-statistic:", z_stat.round(2))
print("p-value:", p_val.round(3))

# Decision rule at 5% significance level
if p_val < 0.05:
    print("We reject H0")
    print("Conclusion: At 5% level of significance, the new program(H1) has improved the pass rate.")
else:
    print("We do not reject H0")
    print("Conclusion: At 5% level of significance, there is no evidence of improvement.")
```

### 5) Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods

```python
# Hypothesis Testing Taken: Independent Two-Sample T-Test (Welch’s Test)

# Question: Is there a difference in average scores between employees trained 
# using Method A and Method B?

# Data collected:
# Method A: n1 = 30, mean1 = 75, std1 = 12
# Method B: n2 = 28, mean2 = 70, std2 = 10

# Hypothesis setup:
# H0: μ1 = μ2 (Null Hypothesis - No difference in average scores)
# H1: μ1 ≠ μ2 (Alternative Hypothesis - There is a difference in average scores)

# Selected 2-Tailed Test:
# To test if the average scores differ between Method A and Method B

import numpy as np
from scipy import stats

# Sample data (we use normal distribution to simulate scores around given mean & std)
np.random.seed(42)  # for reproducibility
methodA_scores = np.random.normal(loc=75, scale=12, size=30)
methodB_scores = np.random.normal(loc=70, scale=10, size=28)

# Perform independent two-sample t-test (Welch’s test by default with equal_var=False)
t_stat, p_val = stats.ttest_ind(methodA_scores, methodB_scores, equal_var=False)

print("t-statistic:", t_stat.round(2))
print("p-value:", p_val.round(3))

# Decision rule at 5% significance level (two-tailed)
if p_val < 0.05:
    print("We reject H0")
    print("Conclusion: At 5% level of significance, there is a difference in average scores(H1) between Method A and Method B, hence H1 is considered")
else:
    print("We do not reject H0")
    print("Conclusion: At 5% level of significance, there is no evidence of a difference in average scores between Method A and Method B. Hence H1 not considered")
```

### 6) The average delivery time from two warehouses is compared using random samples. Warehouse A has a sample of 35 deliveries with a mean time of 48 minutes and a standard deviation of 6 minutes. Warehouse B has a sample of 40 deliveries with a mean time of 52 minutes and a standard deviation of 7 minutes. At the 5% level of significance, test whether the mean delivery times differ

```python
# Hypothesis Testing Taken: Independent Two-Sample T-Test (Welch’s Test)

#Why use Welch’s test?
#When variances are unequal (heteroscedasticity).
#When sample sizes differ.
#It’s more robust and widely recommended in practice, because in real-world data, equal variances are rarely guaranteed.

#Standard two-sample t-test (Student’s t-test): assumes both groups have equal variances.
#Welch’s t-test: relaxes that assumption, allowing variances and sample sizes to differ.

# Question: Is there a difference in average delivery times between Warehouse A and Warehouse B?

# Data collected:
# Warehouse A: n1 = 35, mean1 = 48, std1 = 6
# Warehouse B: n2 = 40, mean2 = 52, std2 = 7

# Hypothesis setup:
# H0: μ1 = μ2 (Null Hypothesis - No difference in mean delivery times)
# H1: μ1 ≠ μ2 (Alternative Hypothesis - Mean delivery times differ)

# Selected 2-Tailed Test:
# To test if the average delivery times differ between Warehouse A and Warehouse B

import numpy as np
from scipy import stats

# Simulating sample data around given mean & std (for demonstration)
np.random.seed(42)  # reproducibility
warehouseA_times = np.random.normal(loc=48, scale=6, size=35)
warehouseB_times = np.random.normal(loc=52, scale=7, size=40)

# Perform independent two-sample t-test (Welch’s test by default with equal_var=False)
t_stat, p_val = stats.ttest_ind(warehouseA_times, warehouseB_times, equal_var=False)

print("t-statistic:", t_stat.round(2))
print("p-value:", p_val.round(2))

# Decision rule at 5% significance level (two-tailed)
if p_val < 0.05:
    print("We reject H0")
    print("Conclusion: At 5% level of significance, there is a difference in mean delivery times between Warehouse A and Warehouse B, then H1 is considered.")
else:
    print("We do not reject H0")
    print("Conclusion: At 5% level of significance, there is no evidence of a difference in mean delivery times between Warehouse A and Warehouse B, Hence H1 is not considered|")
```

### 7) Two versions of a website are tested. Out of 500 users for Version A, 320 users click the signup button. Out of 480 users for Version B, 290 users click the signup button. At the 5% level of significance, test whether the conversion rates differ.

```python
# Hypothesis Testing: Two-Sample Z-Test for Proportions (Direct Function)

# Question: Do the conversion rates differ between Version A and Version B?

# Data:
# Version A: 500 users, 320 clicked signup
# Version B: 480 users, 290 clicked signup

# Hypothesis setup:
# H0: p1 = p2 (Null Hypothesis - Conversion rates are equal)
# H1: p1 ≠ p2 (Alternative Hypothesis - Conversion rates differ)

# Selected 2-Tailed Test:
# To test if the conversion rates differ between Version A and Version B

import numpy as np
from statsmodels.stats.proportion import proportions_ztest

# Sample data
count = np.array([320, 290])   # number of successes
nobs  = np.array([500, 480])   # number of trials

# Perform two-sample z-test for proportions
z_stat, p_value = proportions_ztest(count, nobs, alternative='two-sided')

print("Z Statistic:", z_stat.round(2))
print("P-Value:", p_value.round(3))

# Decision at 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Reject H0: Conversion rates differ significantly.")
else:
    print("Fail to Reject H0: No significant difference in conversion rates.")
```

```python
#Even at 5% Significance difference , there is no signifiant difference in conversion rates, so H1 is not helpful
```

### 8) A vaccine is tested in two regions. In Region A, 180 out of 300 people recover. In Region B, 165 out of 310 people recover. At the 5% level of significance, test whether the recovery rates are significantly different.

```python
# Hypothesis Testing: Two-Sample Z-Test for Proportions

#Why it is Two-Sample Z-Test for Proportions instead of Paired
#Even if the vaccine is the same, as long as the groups are different people in different regions, it is not paired.
#It only becomes a paired test if the same individuals are measured twice 
#(e.g., before vaccination vs after vaccination, or under two different conditions).

# Question: Do the recovery rates differ between Region A and Region B?

# Data:
# Region A: 300 people, 180 recovered
# Region B: 310 people, 165 recovered

# Hypothesis setup:
# H0: p1 = p2 (Null Hypothesis - Recovery rates are equal)
# H1: p1 ≠ p2 (Alternative Hypothesis - Recovery rates differ)

# Selected 2-Tailed Test:
# To test if the recovery rates differ between Region A and Region B

import numpy as np
from statsmodels.stats.proportion import proportions_ztest

# Sample data
count = np.array([180, 165])   # number of recoveries
nobs  = np.array([300, 310])   # total people tested

# Perform two-sample z-test for proportions
z_stat, p_value = proportions_ztest(count, nobs, alternative='two-sided')

print("Z Statistic:", z_stat.round(2))
print("P-Value:", p_value.round(3))

# Decision at 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Reject H0: Recovery rates differ significantly.")
else:
    print("Fail to Reject H0: No significant difference in recovery rates.")
```

```python
#Conclusion : Even at the 5% significance different there is no significant difference in reovery rates of Alternate Hypothesis
```

### 9) A company wants to compare the average productivity of employees trained using three different training programs. Program A has 25 employees with a mean productivity score of 80 and a standard deviation of 8. Program B has 30 employees with a mean of 75 and a standard deviation of 7. Program C has 28 employees with a mean of 82 and a standard deviation of 6. At the 5% level of significance, test whether there is a difference in mean productivity across the programs.

```python
#This problem involves three groups — so instead of a t‑test, we use One‑Way ANOVA (Analysis of Variance) 
#to test whether the mean productivity differs across the three training programs.

# Hypothesis Testing: One-Way ANOVA

# Question: Do the mean productivity scores differ across three training programs?

# Data:
# Program A: n1 = 25, mean1 = 80, std1 = 8
# Program B: n2 = 30, mean2 = 75, std2 = 7
# Program C: n3 = 28, mean3 = 82, std3 = 6

# Hypothesis setup:
# H0: μA = μB = μC (Null Hypothesis - All programs have equal mean productivity)
# H1: At least one program mean differs

# Selected Test:
# One-Way ANOVA (Analysis of Variance)

import numpy as np
from scipy import stats

# Simulated data based on given summary statistics
# (Normally we would use raw data, but here we approximate using normal distributions)
np.random.seed(0)  # for reproducibility

program_A = np.random.normal(loc=80, scale=8, size=25)
program_B = np.random.normal(loc=75, scale=7, size=30)
program_C = np.random.normal(loc=82, scale=6, size=28)

# Perform One-Way ANOVA
f_stat, p_value = stats.f_oneway(program_A, program_B, program_C)

print("F Statistic:", f_stat.round(2))
print("P-Value:", p_value.round(3))

# Decision at 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Reject H0: Mean productivity differs across programs.")
else:
    print("Fail to Reject H0: No significant difference in mean productivity.")
```

```python
# Conclusion: At the 5% level of significance, there is a statistically significant difference in mean productivity across the three training programs.
```

### 10) A company wants to test whether gender and product preference are independent. The survey results are as follows:
*	Male: 60 prefer Product A, 40 prefer Product B
*	Female: 50 prefer Product A, 50 prefer Product B
* At the 5% level of significance, test for independence.


```python
# Hypothesis Testing: Chi-Square Test of Independence

# Question: Are gender and product preference independent?

# Data (Survey Results):
# Male:   60 prefer Product A, 40 prefer Product B
# Female: 50 prefer Product A, 50 prefer Product B

# Hypothesis setup:
# H0: Gender and product preference are independent
# H1: Gender and product preference are not independent

# Selected Test:
# Chi-Square Test of Independence using contingency table

import numpy as np
from scipy.stats import chi2_contingency

# Construct contingency table
# Rows = Gender (Male, Female)
# Columns = Product Preference (A, B)
table = np.array([[60, 40],
                  [50, 50]])

# Perform Chi-Square Test
chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", chi2.round(2))
print("Degrees of Freedom:", dof)
print("Expected Frequencies:\n", expected)
print("P-Value:", p_value.round(2))

# Decision at 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Reject H0: Gender and product preference are associated.")
else:
    print("Fail to Reject H0: Gender and product preference are independent.")
```

```python
#Conclusion: At the 5% level of significance, there is no evidence of association between gender and product preference. 
#Hence gender and product preference appear to be independent.
```

### 11) A  hospital studies whether smoking status is related to disease occurrence. The data collected is as follows:
*	Smokers: 70 have the disease, 30 do not
*	Non-smokers: 40 have the disease, 60 do not
* At the 5% level of significance, test whether smoking status and disease occurrence are independent. 


```python
# Hypothesis Testing: Chi-Square Test of Independence

# Question: Is smoking status related to disease occurrence?

# Data (Survey Results):
# Smokers:    70 have disease, 30 do not
# Non-smokers:40 have disease, 60 do not

# Hypothesis setup:
# H0: Smoking status and disease occurrence are independent
# H1: Smoking status and disease occurrence are not independent

# Selected Test:
# Chi-Square Test of Independence using contingency table

import numpy as np
from scipy.stats import chi2_contingency

# Construct contingency table
# Rows = Smoking Status (Smokers, Non-smokers)
# Columns = Disease Occurrence (Disease, No Disease)
table = np.array([[70, 30],
                  [40, 60]])

# Perform Chi-Square Test
chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", chi2.round(2))
print("Degrees of Freedom:", dof)
print("Expected Frequencies:\n", expected)
print("P-Value:", p_value.round(2))

# Decision at 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Reject H0: Smoking status and disease occurrence are associated.")
else:
    print("Fail to Reject H0: Smoking status and disease occurrence are independent.")
```

### 12) 12.	A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
*	Face 1: 18
*	Face 2: 22
*	Face 3: 20
*	Face 4: 25
*	Face 5: 17
*	Face 6: 18
* At the 5% level of significance, test whether the die is fair.


```python
# Hypothesis Testing: Chi-Square Goodness-of-Fit Test

#Since we’re checking whether the observed dice outcomes match the expected distribution of a fair die.
#So, choosen Chi-Square Goodness-of-Fit Test

# Question: Is the die fair?

# Data (Observed Frequencies from 120 rolls):
# Face 1: 18
# Face 2: 22
# Face 3: 20
# Face 4: 25
# Face 5: 17
# Face 6: 18

# Hypothesis setup:
# H0: The die is fair (all faces equally likely)
# H1: The die is not fair

# Selected Test:
# Chi-Square Goodness-of-Fit Test

import numpy as np
from scipy.stats import chisquare

# Observed frequencies
observed = np.array([18, 22, 20, 25, 17, 18])

# Expected frequencies (fair die → equal probability)
expected = np.array([120/6] * 6)

# Perform Chi-Square Goodness-of-Fit Test
chi2_stat, p_value = chisquare(f_obs=observed, f_exp=expected)

print("Chi-Square Statistic:", chi2_stat.round(2))
print("Expected Frequencies:", expected)
print("P-Value:", p_value.round(3))

# Decision at 5% significance level
alpha = 0.05
if p_value < alpha:
    print("Reject H0: The die is not fair.")
else:
    print("Fail to Reject H0: No evidence against fairness of the die.")
```

```python
#Conclusion: Even at 5% of significance difference, there is no guarantee that die is always fair
```
