# 1. Data Cleaning Challenge on Sales Data.

```python
import pandas as pd
sales_data1 = pd.read_csv("sales_transactions_2017.csv")
sales_data2 = pd.read_csv("sales_transactions_2018.csv")
sales_data3 = pd.read_csv("sales_transactions_2019.csv")

total_sales = pd.concat([sales_data1,sales_data2,sales_data3],ignore_index=True)

total_sales.to_csv("combined_sales.csv",index = False)

```

```python
#Descirbe to find mean median mode
total_sales.describe()
pd.set_option("display.max_rows",1000)
```

```python
#info
total_sales.info()
```

```python
#Find duplicate entry
total_sales.duplicated().sum()
```

### 2. Remove Duplicates

```python
total_sales = total_sales.drop_duplicates()
total_sales.duplicated().sum()
```

### 3. Handle Missing values

```python
total_sales.isna().sum()
```

#### 3.1 Fill current date where date cell is nan

```python
#place current date where date cell has nan
total_sales.loc[total_sales["Date"].isna(),"Date"] = "28/07/2026"
total_sales
```

#### 3.2 Write incremented voucher number where voucher cell is nan

```python
# Create a sequence starting at 101 for missing values
start_value = 101
missing_count = total_sales['Voucher'].isna().sum()

# Generate the series
voucher_series = range(start_value, start_value + missing_count)

# Fill NaN with the series
total_sales.loc[total_sales['Voucher'].isna(), 'Voucher'] = list(voucher_series)

total_sales
```

### 3.3 Remove the observations where party or product is having nan as both are categorical data can not define the suitable name


```python
total_sales = total_sales.dropna(subset = ["Party","Product"])
total_sales
```

# Task 2 Merge Function 1 Example

```python
import pandas as pd

df1 = pd.DataFrame({"id": [1, 2, 3], "name": ["Kannan", "Arun", "Priya"]})
df2 = pd.DataFrame({"id": [2, 3, 4], "age": [30, 25, 40]})

merged = pd.merge(df1, df2, on="id")
print(merged)
## whatever id common 2nd datafram columns are merged, even it control further level also
#we can control this with the how parameter:
#how="inner" → only matching IDs (default).
#how="left" → all IDs from df1, match if possible.
#how="right" → all IDs from df2, match if possible.
#how="outer" → all IDs from both, fill missing with NaN.

print("----------------------------------------------------")
mergd2 = pd.merge(df1, df2, on="id", how="outer")
print(mergd2)
```

# 3. List Comprehension 2 examples

list comprehensions are very handy with pandas too. They’re often used when you want to quickly build a new column or transform data without writing long loops.


### 1. Create a New Column with Transformation

```python
import pandas as pd

df = pd.DataFrame({"name": ["Kannan", "Arun", "Priya"], "age": [25, 30, 28]})

# Add a new column with age doubled
df["double_age"] = [age * 2 for age in df["age"]]

print(df)
```

### 2: Conditional Transformation

```python
# Label people as 'Young' if age < 28 else 'Adult'
df["category"] = ["Young" if age < 28 else "Adult" for age in df["age"]]

print(df)
```

# 4. Lambda Function 2 examples with Map, Filter, Reduce

```python
# Example 1:
# A lambda that adds 10 to a number
add_ten = lambda x: x + 10

print(add_ten(5))   # 15
print(add_ten(20))  # 30
```

```python
#Example 2
import pandas as pd

df = pd.DataFrame({"name": ["Kannan", "Arun", "Priya"], "age": [25, 30, 28]})

# Add 5 years to each age using lambda + apply
df["future_age"] = df["age"].apply(lambda x: x + 5)

print(df)
```

```python
#Example 3
# Categorize based on age
df["category"] = df["age"].apply(lambda x: "Young" if x < 28 else "Adult")

print(df)
```

```python
#Example 4 with Map and filter
numbers = [1, 2, 3, 4, 5, 6]

# Double each number using map
doubled = list(map(lambda x: x * 2, numbers)) # map applies the lambda to every element.
print("Doubled:", doubled)  # [2, 4, 6, 8, 10, 12]

# Filter even numbers using filter
evens = list(filter(lambda x: x % 2 == 0, numbers)) # filter keeps only those elements where the lambda returns True.
print("Evens:", evens)  # [2, 4, 6]
```

```python
# Example 5 with reduce
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# Sum of all numbers using reduce
total = reduce(lambda x, y: x + y, numbers)
print("Sum:", total)  # 15

# Product of all numbers using reduce
product = reduce(lambda x, y: x * y, numbers)
print("Product:", product)  # 120
```
