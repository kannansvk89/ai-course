### 3. Write a program that handles division by zero using try-except.

```python
while True:
    try:
        quotient = float(input("Enter a quotient : "))
        denometer = float(input("Enter a denometer: "))
       
        result = quotient/denometer
        print("Answer:",result)
        break;
    
    except ValueError as e:
        print("You entered invalid number, Please try again")
    except ZeroDivisionError as e:
        print("Quotient can not be divided by zero Denometer")
```

### 4.Given a string, count how many vowels it has.

```python
input_text = input("Enter a String")
vowels = set()
for c in input_text:
    if c in "aeiou":
        vowels.add(c)
print("In the enterted string -->",input_text,"and Vowels are",vowels)
```

### 5. Reverse a list without using built-in functions.

```python
fruit_list = ["Apple","Mangao","Banana","Orange"]
list_len = len(fruit_list)
iterate = int(list_len/2)

for i in range(iterate):
     temp = fruit_list[i]
     fruit_list[i] = fruit_list[(list_len-i)-(list_len+1)] ## Acces last element using negative index
     fruit_list[(list_len-i)-(list_len+1)]  = temp
print(fruit_list)


```
