# Data Visualization

**Types of charts**
* Histogram
* Bar plot
* Count plot
* Pie chart
* Pair plot
* Heatmap
* Line chart
* Stacked Chart
* Scatter plot


```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
mtcars_data = pd.read_csv("../csv/mt_cars.csv")
mtcars_data.head()
mtcars_data.isna().sum() # Data is aleady cleaned
```

```python
mtcars_data.head();
mtcars_data.describe().round(2)
```

```python
mtcars_data.hp.nunique()
```

####  1. Historgram  - Shows how fuel efficiency is distributed across cars.

```python
#Histogram chart speaks about the distribution of one continous data
# X-Axis - Continous data
plt.hist(mtcars_data.mpg,color="skyblue", edgecolor="black")
plt.title("Histogram of MPG")
plt.xlabel("Miles per Gallon")
plt.ylabel("Frequency")
plt.show()
```

```python
#Inshights
#Most of the cars are giving mileage between 15 to 25
#Note: With histogram chart we can tell the insight figue with in a range
mtcars_data.groupby("mpg").size()
```

```python
# how HP is distributed across the cars
plt.hist(mtcars_data.hp,color="skyblue", edgecolor="black")
plt.title("Histogram of HP")
plt.xlabel("Horse Power")
plt.ylabel("Frequency")
plt.show()
```

#### 2: Bar Plot (Average MPG by Cylinder)

```python
#x -axis discrete data, y- axis continous data
sns.barplot(data=mtcars_data, x="cyl", y="mpg", estimator="mean")
plt.title("Average MPG by Cylinder")
plt.show()
```

```python
#Insights
#4-cylinder cars are most fuel-efficient.
#8-cylinder cars consume more fuel.
```

#### 3. Pairplot

```python
sns.pairplot(data = mtcars_data)
```

#### 4. Count Plot (Number of Cars by Gear)

```python
#to understand the distibution of one discrete data, countplot is useful
#Y-Axis - discrete data
sns.countplot(data=mtcars_data, y="gear")
plt.title("Count of Cars by Gear")
plt.show()
```

```python
#insights
#There are five cars are exist with 5 gear
#There 13 cars are exist with 4 gear
#Note: compare to x axis, analyzing with y is bett 
```

#### 5. Piechart Not common in data science, but useful for proportions.

```python
mtcars_data.value_counts("cyl").plot.pie(autopct="%1.1f%%", colors=["lightblue","lightgreen","orange"],labels=["8 cyl","6 cyl","4 cyl"])
plt.title("Cylinder Proportion")
plt.ylabel("")
plt.show()
```

#### 6. ScatterPlot - find relation between two continous data

```python
#X-Axis continous data y-Axis discrete data
plt.scatter(x="wt",y="mpg",data = mtcars_data)
```

```python
#insights
# 1) if weight is increasing then average mileage is reducing
# 2) Overall the car weight is between 2.5 to 4.0 is gaining good mileage
```

```python
sns.scatterplot(data=mtcars_data, x="hp", y="mpg", hue="cyl", size="wt",)
plt.title("Horsepower vs MPG")
plt.show()

#X-axis: Horsepower (hp)
#Y-axis: Miles per gallon (mpg)
#Color: Number of cylinders (cyl)
#Size: Weight (wt)
```

#### 7. HeatMap - find the strength of relationship between 2 features

```python
mtcars_data.corr().round(2)
```

```python
del mtcars_data["model"]
```

```python
sns.heatmap(mtcars_data.corr(),annot=True)
```

```python
#inights
# 1)Strong negative correlation: hp vs mpg.
# 2)Positive correlation: wt vs hp.
```

#### 8. LindChart - Analyze betweeen contionuse and discrete data

```python
plt.plot(mtcars_data.index, mtcars_data["mpg"], marker="o")
plt.title("Line Plot of MPG by Car Index")
plt.xlabel("Car Index")
plt.ylabel("MPG")
plt.show()
```

#### 9. Stacked chart

```python
pd.crosstab(mtcars_data["cyl"], mtcars_data["gear"]).plot(kind="bar", stacked=True)
plt.title("Stacked Bar: Cylinders vs Gear")
plt.xlabel("Cylinders")
plt.ylabel("Count")
plt.show()
```

```python
#Insights
#Shows how gear types are distributed within cylinder groups.
```

#### 10. KDE Chart - Kernel Density Estimation.
A smoothed version of the histogram.

```python
sns.kdeplot(mtcars_data["mpg"], fill=True, color="green")
plt.title("KDE Plot of MPG")
plt.show()
```

```python
#Insights
#Helps visualize the probability density of MPG values.
```

```python
sns.violinplot(data=mtcars_data, x="cyl", y="mpg", hue="cyl", palette="muted", legend=False)
plt.title("Violin Plot of MPG by Cylinder")
plt.show()
```
