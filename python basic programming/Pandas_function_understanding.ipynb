```python
order_data.groupby(by="item_name")["item_price"].sum().sort_values(ascending=False).head()
```

## 2. Total revenue of the 5 item sold

```python
order_data.groupby(by="item_name")["item_price"].sum().sort_values(ascending=False).head().sum()
```

## 3. find which items have sold more than 100 quantities, you need to group by

```python
qty = order_data.groupby(by="item_name")["quantity"].sum()
qty.loc[qty>100]
```
