```python
import pandas as pd
```

```python
insurance_data = pd.read_csv("insurance.csv")
```

## 1. Groupby

```python
insurance_data.describe(include="all")
```

#### 1.2 Pull up the average insurance charge based on region

```python
insurance_data.isna().sum()
```

#### 1.3 Pull up the average insurance charges based on region and sex

 ##### Discrete and contiouns data
  * Discrete - Just countable/ Can't segemented
  * Contious - Measurable/Have units /can also segemented

```python
insurance_data.groupby(by=["region","sex"])["charges"].mean().round(2).sort_values()
```

### 2. Pivot Table
##### #Presentable comparing 2 categorical features vs 1 contious data

```python
#Pivot Table
pd.pivot_table(data =insurance_data,values = "charges",index="region",columns=["sex"]).round()
```

```python
pd.pivot_table(data =insurance_data,values = "charges",index="region",columns=["sex","children"]).round()
#Presentable comparing 2 categorical features vs 1 contious data
```

### 3. Aggregate

#### 3.1 Find out the average, max, min, sum of charges basedon region

```python
insurance_data.groupby(by="region")["charges"].aggregate(func = ["min","max","mean","sum"]).round(2)
```

### 4. Cross Tab

#### 4.1 how many smokers and non-smokers based on region

```python
#To compute 2 categorical features we use the crosstab
# To find total frequencies between two categorical data
pd.crosstab(index = insurance_data["region"],columns = insurance_data["smoker"],margins=True)
```

### 5. Concat

```python
insurance_data.age
```

### 6. Apply

#### 6.1 Add new column eligible_criteria based on age

```python
def get_age(x):
    if x > 20:
        return "Eligble"
    else:
        return "Not Eligible"
    

insurance_data["Eligible_Criteria"] = insurance_data["age"].apply(get_age)

insurance_data
```

### 7. Lambda

```python
insurance_data["Eligible_Criteria"] = insurance_data["age"].apply(lambda x:1 if x > 20 else 0)

insurance_data
```
