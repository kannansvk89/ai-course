### 1. Encapsulation - Hiding data to outside

```python
class Employee:
  
    def __init__(self,name,dept,salary):
        self.name = name
        self.dept = dept
        self.__salary = salary ## define private members
    
    def displayEmpDetails(self):
       print(f"Name: {self.name}, Dept: {self.dept}, Salary: ${self.__getSalary()}")

    def __getSalary(self):
        return self.__salary

emp1 = Employee("kannan","AI","5500000")
emp1.displayEmpDetails()



```

### 2. Inheritance

```python
class Employee:

    def __init__(self, name, dept, salary):
        self.name = name
        self.dept = dept
        self.__salary = salary

    def displayEmpDetails(self):
        print(f"Name: {self.name}, Dept: {self.dept}, Salary: ${self.__salary}")


class Manager(Employee):  ## inherits the members of employee class as it applicable for manager role

    def manageTeam(self):
        print(f"{self.name} is managing the {self.dept} team.")

class CFO(Employee):

    def manageCompany(self):
        print(f"{self.name} is managing the {self.dept} team.")


manager1 = Manager("Kannan", "AI", 5500000)
manager1.displayEmpDetails()
manager1.manageTeam()

cfo = CFO("Kanmani", "Company", 6500000)
cfo.manageCompany()
```

#### 3. Astraction  -  abstraction is about hiding implementation details and exposing only the essential features.

```python
from abc import ABC, abstractmethod ## Need to import the ABC class and abstractmethod decorator inorder to implement abstraction
class Employee(ABC):
    def __init__(self,name,dept):
        self.name = name
        self.dept = dept

    def displayEmpDetails(self):
        print(f"Name: {self.name}, Dept: {self.dept}")
         
    @abstractmethod
    def work(self):
        pass

class Manager(Employee):
    def test():
        pass

    def work(self):
        print("--->Manager managing the teams")
        
class Developer(Employee):
    def work(self):
        print("Developer is analyze and writting code")

manager = Manager("Narpaviy","Manager")
manager.displayEmpDetails()
manager.work() ###--> Polymorphism Method Overriding

developer = Developer("Makizhini","TechnicalLead")
developer.work() ###--> Polymorphism - Method overriding
```
